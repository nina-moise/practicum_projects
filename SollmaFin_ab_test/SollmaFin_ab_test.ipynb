#  Исследование результатов А/В-теста и поиск инсайтов

**Дата:** 20.09.2025

SollmaFin — международное мобильное приложение для онлайн-торговли финансовыми активами. Пользователи могут инвестировать деньги в акции, валюту, криптовалюту, биржевые фонды (ETF) и другие активы. Целевая аудитория приложения — начинающие инвесторы. Получить доступ к финансовым рынкам можно со смартфона либо через веб-версию. 

Приложение ориентировано на рынок Латинской Америки с акцентом на четыре страны: Мексику, Бразилию, Колумбию и Аргентину — там быстро растёт интерес к финтеху и инвестициям, а экономики очень разные. 

Появилась гипотеза о недостаточной финансовой грамотности пользователей: они не понимают, как работать с активами разной степени риска. Клиенты покупают активы с высоким риском и теряют деньги, что снижает вовлечённость.
Команда продукта решила обновить онбординг пользователей и добавить в него детальную информацию о различиях финансовых активов и связанных с ними рисках. Однако появилось опасение, что углублённый онбординг отпугнёт клиентов от пополнения депозита и более рискованных вложений. 

Чтобы разобраться, команда решила провести A/B-эксперимент. 

### А/В-эксперимент

В эксперименте участвовали новые пользователи, которые зарегистрировались в приложении со 2 по 15 июня 2025 года. 
Пользователей случайным образом разделили на две равные группы: 
- Контрольная группа проходила стандартный онбординг без обязательного обучения.
- Тестовая группа проходила обновлённый онбординг с подробной информацией об активах и связанных с ними рисках. 

После разделения активность пользователей анализировали в течение недели.

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита. 

Чтобы увидеть полную картину, недостаточно следить за метриками конверсии. Команда выбрала для эксперимента такой набор метрик: 
- Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию).
- Барьерная метрика — конверсия из регистрации в первый депозит.
- Вспомогательная метрика 1 — конверсия из первого депозита во второй.
- Вспомогательная метрика 2 — средняя сумма всех депозитов на пользователя, который открыл хотя бы один депозит.

Ожидалось, что ключевая и барьерная метрики не упадут, а вспомогательные покажут значительный рост.

#### Динамика привлечения новых пользователей в приложение по дням

**Задание** Анализ новых пользователей.
- Изучим динамику привлечения новых пользователей в приложение.

```python
# Считаем количество новых пользователей по дате первого появления в приложении
df_users_day = df_hist.groupby('first_dt')['user_id'].nunique().reset_index()
df_users_day = df_users_day.rename(columns={'user_id' : 'n_users'})

# визуализируем
plt.subplots(figsize=(12,5)) 

plt.plot(df_users_day['first_dt'], df_users_day['n_users'])

plt.title('Динамика привлечения новых пользователей по дням') 
plt.xlabel('Дата') 
plt.ylabel('Количество') 
plt.xticks(rotation=45)  
plt.grid() 
plt.show() 
```

Новые пользователи приходят каждый день, но неравномерно. Данные по новым пользователям сильно скачут из дня в день, и чтобы лучше понять общую картину, используем **скользящее среднее**. Оно сгладит резкие колебания и поможет увидеть, как меняется количество новых пользователей в целом, без резких провалов и пиков.

```python
# Сортируем данные по дате
df_users_day = df_users_day.sort_values('first_dt')

# Считаем количество дней 
n_days=df_users_day['first_dt'].nunique()

# Считаем скользящее среднее за n_days дней
df_users_day['n_users_ma'] = df_users_day['n_users'].rolling(window=n_days, min_periods=1).mean()

# визуализируем скользящее среднее
plt.subplots(figsize=(12,5)) 
plt.plot(df_users_day['first_dt'], df_users_day['n_users_ma'])

plt.title('Динамика привлечения новых пользователей по дням (скользящее среднее)') 
plt.xlabel('Дата') 
plt.ylabel('Количество') 
plt.xticks(rotation=45)  
plt.grid() 
plt.show() 
```

**Вывод:** Привлечение новых пользователей было на пике в самом начале и стабильно во время всего периода.

#### Анализ классической воронки

**Задача** Анализ воронок событий.
- Построим классическую и пошаговую воронки;
- Охарактеризуем пользовательский путь в приложении и определим узкие места воронки. 

```python
# Задаем порядок шагов в воронке
funnel_steps = ['install / open_web','introduction','registration','main_page','onboarding_complete',\
                'first_deposit','asset_purchase','second_deposit']
```

Видим, что в воронке 8 последовательных событий.

```python
# Считаем число уникальных пользователей на каждом шаге
funnel_counts = df_hist.groupby('event_name')['user_id'].nunique().reindex(funnel_steps)
```

```python
# Собираем таблицу воронки
funnel_df = pd.DataFrame({
   'step': pd.Series(funnel_counts.index),
   'users': pd.Series(funnel_counts.values)})
```

```python
# Добавляем конверсии и округляем до 1 знака после запятой
funnel_df['conversion_from_first_%'] = round((funnel_df['users'] / funnel_df['users'].iloc[0]) *100, 1)
```

```python
display(funnel_df.head(8))
```

```python
# Визуализируем классическую воронку
plt.figure(figsize=(10, 6))

bars = plt.bar(funnel_df['step'], funnel_df['conversion_from_first_%'], color='skyblue')

for bar in bars:
    yval = bar.get_height() # высота столбца=значение
    plt.text(bar.get_x() + bar.get_width()/2, yval, f'{yval}%', verticalalignment='bottom', horizontalalignment='center')
    # комментарий для себя: можно также писать va= ha=

    plt.title('Классическая воронка')
plt.ylabel('Конверсия от первого шага (%)')
plt.xlabel('Этапы воронки')
plt.xticks(rotation=45)  
plt.show()
```

**Вывод:** По классической воронке количество пользователей уменьшается неравномерно. До последнего шага воронки - второго депозита - доходит только 13,2% пользователей от первоначального количества. Значительное уменьшение наблюдается на трех шагах:
- шаг из "завершения онбординга" в "первый депозит" - теряется 16,2%
- шаг из "первого депозита" в "покупку актива" - теряется 30,0%
- шаг из "покупки актива" во "второй депозит" - теряется 24,3% пользователей.

#### Анализ пошаговой воронки (step-by-step)

```python
# Рассчитываем пошаговую воронку
funnel_df['conversion_from_previous_%'] = round(funnel_df['users'] / funnel_df['users'].shift(1) * 100,1)
```

```python
# Визуализируем воронку step-by-step 
funnel_df = funnel_df.iloc[1: ] # чтобы не отображать самый первый шаг с Nan

plt.figure(figsize=(10, 6))

bars = plt.bar(funnel_df['step'], funnel_df['conversion_from_previous_%'], color='skyblue')

for bar in bars:
    yval = bar.get_height() # высота столбца=значение
    plt.text(bar.get_x() + bar.get_width()/2, yval, f'{yval}%', verticalalignment='bottom', horizontalalignment='center')

plt.title('Step-by-step воронка: конверсия от предыдущего шага')
plt.ylabel('Конверсия от предыдущего шага (%)')
plt.xticks(rotation=45)  
plt.xlabel('Этапы воронки')
plt.show()
```

**Вывод:** По воронке step-by-step больше всего теряется пользователей на этапе перехода в "покупку актива" - 25%.
Также при переходе ко "второму депозиту" теряется 20,5% пользователей.

Также обращает внимание, что 7,1% пользователей теряется после просмотра "введения" при переходе к "регистрации". Возможно, проблема возникает из-за неудачного интерфейса.

#### Анализ уровня риска актива на открытие второго депозита

**Задача** Влияние уровня риска актива на открытие второго депозита.
- Изучим, как уровень риска первого актива влияет на конверсию во второй депозит.  

Столбец 'уровень риска' 'risk_level' заполняется только для события 'asset_purchase', после совершения события 'first_deposit'. Поэтому сформируем цепочку событий для пользователя - 'asset_purchase'- 'second_deposit'. 

```python
# Выбираем пользователей, которые совершили покупку актива после 1 депозита
df_users_asset = df_hist.loc[df_hist['event_name'] == 'asset_purchase']
```

```python
# Выбираем пользователей, которые открыли 2 депозит
df_users_2_deposit = df_hist.loc[df_hist['event_name'] == 'second_deposit']
```

```python
# Объединим данные по пользователю 
df_user_risk = pd.merge(df_users_asset[['user_id','event_name','risk_level']], 
              df_users_2_deposit[['user_id', 'event_name','amount']], on='user_id', how='left' )
```

```python
# Сгурппируем по степени риска и посчитаем количество уникальных пользователей с 1 депозитом
df_risk_1 = df_user_risk.groupby('risk_level')['user_id'].nunique().reset_index()
```

```python
# Переименуем столбец
df_risk_1 = df_risk_1.rename(columns = {'user_id': 'count_dep1'})
```

```python
# Выберем пользователей с событием "второй депозит",
# сгруппируем по степени риска и посчитаем уникальное количество пользователей
df_risk_2 = df_user_risk.loc[df_user_risk['event_name_y'] == 'second_deposit'].groupby('risk_level')\
            ['user_id'].nunique().reset_index()
```

```python
# Переименуем столбец
df_risk_2 = df_risk_2.rename(columns = {'user_id': 'count_dep2'})
```

```python
# Объединим посчитанные данные
df_risk = pd.merge(df_risk_1, df_risk_2, on='risk_level', how='inner')
```

#### Классическая и пошаговая воронки: сравнение двух групп 

**Задача:** Сравнение воронок событий.
- Сравним пользовательский путь новых пользователей в тестовой и контрольной группах.
- Построим классическую и последовательную воронки, выполним визуализацию и сделаем промежуточный вывод. 

```python
# Задаем порядок шагов в воронке
funnel_steps = ['install / open_web','introduction','registration','main_page','onboarding_complete',\
                'first_deposit','asset_purchase','second_deposit']
```

```python
# Разделим данные по тестовым группам
control_df = df_exp.loc[df_exp['group'] == 'control']
test_df = df_exp.loc[df_exp['group'] == 'test']
```

Считаем показатели воронок для контрольной группы

```python
# Для контрольной группы - считаем количество клиентов на каждом шаге
counts_control_agg = control_df.groupby('event_name')['user_id'].nunique().reindex(funnel_steps)
```

```python
# Классическая воронка: для контрольной группы - cчитаем конверсию от первого шага 
control_classic_funnel = ((counts_control_agg.values / counts_control_agg.values[0]) * 100).round(1)
```

```python
# Пошаговая воронка: для контрольной группы - cчитаем конверсию от предыдущего шага 
control_step_funnel = (counts_control_agg.values / counts_control_agg.shift(1) * 100).round(1)
```

Считаем показатели воронок для тестовой группы

```python
# Для тестовой группы - считаем количество клиентов на каждом шаге
counts_test_agg = test_df.groupby('event_name')['user_id'].nunique().reindex(funnel_steps)
```

```python
# Классическая воронка: для тестовой группы - cчитаем конверсию от первого шага 
test_classic_funnel = ((counts_test_agg.values / counts_test_agg.values[0]) * 100).round(1)
```

```python
# Пошаговая воронка: для тестовой группы - cчитаем конверсию от предыдущего шага 
test_step_funnel = (counts_test_agg.values / counts_test_agg.shift(1) * 100).round(1)
```

```python
# Соберём данные в датафрейм 
funnel_compare = pd.DataFrame({
   'control_classic': control_classic_funnel,
   'control_step': control_step_funnel,
   'test_classic': test_classic_funnel,
   'test_step': test_step_funnel})
```

```python
display(funnel_compare.head(8))
```

```python
# Визуализируем классическую воронку в сравнении по тестовым группам
# Данные для диаграммы
категории = funnel_compare.index
значения1 = funnel_compare['control_classic']
значения2 = funnel_compare['test_classic']

# Ширина столбцов
ширина_столбца = 0.35

# Создание списка позиций для столбцов на оси X
позиции = np.arange(len(funnel_compare))

# Создание объектов фигуры и осей
fig, ax = plt.subplots(figsize=(10, 6))

# Построение столбчатых диаграмм
прямоугольник1 = ax.bar(позиции - ширина_столбца/2, значения1, ширина_столбца, label='control', color='#ff999a')
прямоугольник2 = ax.bar(позиции + ширина_столбца/2, значения2, ширина_столбца, label='test', color='#9acc99')

# Настройка оси X
ax.set_xticks(позиции)
ax.set_xticklabels(категории)
ax.set_ylabel('Конверсия от первого шага (%)')
ax.set_title('Сравнение этапов классической воронки для 2-х групп эксперимента')
ax.legend()

# Добавление подписей над столбцами
def add_labels(rects, ax):
    for rect in rects:
        height = rect.get_height()
        ax.text(rect.get_x() + rect.get_width()/2., height + 2,
                f'{height}',
                ha='center', va='bottom')

add_labels(прямоугольник1, ax)
add_labels(прямоугольник2, ax)

# Отображение диаграммы
plt.tight_layout()
plt.xticks(rotation=45)  
plt.show()
```

**Вывод:** По классической воронке в тестовой группе значительно выросло количество пользователей на последнем шаге воронки - 2 депозит - 21,3% против 13,5% в контрольной (разница 7,8%). Также в тестовой группе на 6,7% выросло количество на шаге "покупка актива" (тестовая - 43,3%, контрольная - 36,6%).

Но на 8% в тестовой группе упало количество пользователей при переходе к шагу "завершение онбординга". И на 1,2% в тестовой группе снизилось количество инвестирующих в 1 депозит. Это связано с изменениями в новом интерфейсе онбординга.

```python
# Визуализируем пошаговую воронку в сравнении по тестовым группам - не берем первый шаг, т.к. не с чем сравнивать
# Данные для диаграммы
категории = funnel_compare.index[1:]
значения1 = funnel_compare['control_step'].iloc[1:]
значения2 = funnel_compare['test_step'].iloc[1:]

# Ширина столбцов
ширина_столбца = 0.35

# Создание списка позиций для столбцов на оси X
позиции = np.arange(len(funnel_compare)-1)

# Создание объектов фигуры и осей
fig, ax = plt.subplots(figsize=(10, 6))

# Построение столбчатых диаграмм
прямоугольник1 = ax.bar(позиции - ширина_столбца/2, значения1, ширина_столбца, label='control', color='#ff999a')
прямоугольник2 = ax.bar(позиции + ширина_столбца/2, значения2, ширина_столбца, label='test', color='#9acc99')

# Настройка оси X
ax.set_xticks(позиции)
ax.set_xticklabels(категории)
ax.set_ylabel('Конверсия от предыдущего шага (%)')
ax.set_title('Сравнение этапов воронки Step-by-step для 2-х групп эксперимента')
ax.legend()

# Добавление подписей над столбцами
def add_labels(rects, ax):
    for rect in rects:
        height = rect.get_height()
        ax.text(rect.get_x() + rect.get_width()/2., height + 2,
                f'{height}',
                ha='center', va='bottom')

add_labels(прямоугольник1, ax)
add_labels(прямоугольник2, ax)

# Отображение диаграммы
plt.tight_layout()
plt.xticks(rotation=45)  
plt.show()
```

Проверим статистическую значимость полученных данных с помощью Z-теста, т.к. оцениваются доли.

Формулируем гипотезы:
- H0: Доля пользователей, инвестирующих во второй депозит в контрольной и тестовой группах не различается.
- H1: Доля пользователей, инвестирующих во второй депозит в тестовой группе больше, чем в контрольной (правосторонняя гипотеза).

```python
# Считаем размеры групп 
n_a = df_exp.loc[df_exp['group'] == 'control']['user_id'].nunique()
n_b = df_exp.loc[df_exp['group'] == 'test']['user_id'].nunique()
```

```python
# Считаем количество успехов в группах
m_a = df_exp.loc[(df_exp['group'] == 'control') & (df_exp['event_name'] == 'second_deposit')]['user_id'].nunique()
m_b = df_exp.loc[(df_exp['group'] == 'test') & (df_exp['event_name'] == 'second_deposit')]['user_id'].nunique() 
```

```python
# Считаем долю успехов для каждой группы                 
p_a, p_b = m_a/n_a, m_b/n_b 
                 
# Проверяем предпосылку о достаточном количестве данных
if (p_a*n_a > 10) and ((1-p_a)*n_a > 10) and (p_b*n_b > 10) and ((1-p_b)*n_b > 10):
    display('Предпосылка о достаточном количестве данных выполняется!')
else:
    display('Предпосылка о достаточном количестве данных НЕ выполняется!')
```

```python
# Применяем Z-тест пропорций
alpha = 0.05 ## уровень значимости

# smaller - pa<pb
stat_ztest, p_value_ztest = proportions_ztest([m_a, m_b],[n_a, n_b],alternative='smaller')

display(f'pvalue = {p_value_ztest}') 

if p_value_ztest < alpha:
    display('Отвергаем H0. Принимаем альтернативную гипотезу: Доля пользователей, \
инвестирующих во второй депозит в тестовой группе больше, чем в контрольной.')
else:
    display('Подтверждаем H0. Доля пользователей, инвестирующих во второй депозит \
в контрольной и тестовой группах не различается.')
```

**Вывод:** Получены статистически значимые данные, что в тестовой группе, по сравнению с контрольной, увеличилась конверсия во второй депозит для всех степеней риска. Самое значительное изменение - для высокой степени риска. Здесь конверсия выросла на 18,34%.

Рассчитаем и визуализируем кумулятивное значение ключевой метрики для тестовых групп за весь период эксперимента. Для этого сформируем датасет:
- group (группа эксперимента),
- data (дата эксперимента)
- значение ключевой метрики до даты эксперимента

```python
# Формируем список дат для цикла и сортируем его
date_list = df_exp['event_ts'].dt.date.unique()
date_list.sort()
```

```python
records = []
for current_date in date_list:
    # Выбираем все события до текущей даты включительно
    df_exp_to_date = df_exp.loc[df_exp['event_ts'].dt.date <= current_date]
    
    # Группируем по пользователю и тестовой группе и вычисляем сумму всех депозитов на пользователя
    df_ = df_exp_to_date.groupby(['group','user_id']).agg(apru = ('amount','sum')).reset_index()
    # Группируем по тестовой группе и считаем метрику
    df_ = df_.groupby('group')['apru'].mean().reset_index()
    df_['date'] = current_date
    records.append(df_)
    
    # Собираем всё в итоговый датафрейм
    report = pd.concat(records)
```

```python
# Строим график
color_list = ['#ff999a','#9acc99']
plt.figure(figsize=(10, 6))
for group, color_ in zip(report['group'].unique(), range(0, len(color_list))):
    group_data = report.loc[report['group'] == group]
    plt.plot(group_data['date'], group_data['apru'] , color=color_list[color_], marker='o', label=group)
   
plt.title('Кумулятивное значение средней суммы всех депозитов на одного пользователя (APRU)')
plt.xlabel('Дата')
plt.ylabel('APRU')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Вывод:** По графику видно, как происходило накопление коллективного пользовательского опыта в тестовой группе - до 12 июня ключевая метрика в контрольной группе была выше, чем тестовой. Но, после 12 июня, метрика в тестовой группе стала выше и оставалась выше до конца эксперимента. Также стоит обратить снимание на "ступеньки" - то, что они появились, может намекать нам на то, что пользователи начали делать вторые, крупные депозиты. А ступенька после 16 числа - именно в этот день завершился набор новых пользователей, и далее в выборке остались только активные пользователи.

Посмотрим на динамику p-value по дням эксперимента.

```python
pval_records = []

for current_date in date_list:
    # Выбираем все события до текущей даты включительно
    df_exp_to_date = df_exp.loc[df_exp['event_ts'].dt.date <= current_date]
   
    group_control = df_exp_to_date.loc[df_exp_to_date['group'] =='control'].groupby('user_id')['amount'].mean().reset_index()  
    group_test = df_exp_to_date.loc[df_exp_to_date['group'] =='test'].groupby('user_id')['amount'].mean().reset_index()  

    # Заменим Nan для тех пользователей, котороые не совершали депозитов, на 0
    group_control['amount'] = group_control['amount'].fillna(0)
    group_test['amount'] = group_test['amount'].fillna(0)
    
    # Тестируем
    result = st.ttest_ind(group_control['amount'], group_test['amount'], equal_var=False, alternative = 'greater')

    pval_records.append({'date': current_date,'p_value': result.pvalue})
       
# Собираем всё в итоговый датафрейм
df_pvalue = pd.DataFrame(pval_records)
```

```python
# Строим график
plt.figure(figsize=(10, 6))
plt.plot(df_pvalue['date'], df_pvalue['p_value'], marker='o', label='p_value' )
   
plt.title('Динамика p-value по дням эксперимента')
plt.xlabel('Дата')
plt.ylabel('P_value')
plt.grid(True)

plt.axhline(y=0.05, linestyle='--', color='black', label='alpha = 0.05') 
plt.legend()
plt.tight_layout()
plt.show()
```

**Вывод:** Динамика p_value по дням отражает (несмотря на колебания в начале) устойчивое значение ниже уровня значимости с первых дней до конца эксперимента, что говорит о стабильном результате.

**Вывод:** В расчетах по данным эксперимента барьерная метрика - конверсия из регистрации в первый депозит- в тестовой группе упала  на 2,1%. Но статистический тест говорит о том, что различий между тестовыми группами по этому показателю не наблюдается. Значит, падение барьерной метрики можно считать случайностью, и не подтвердилась гипотеза риска о том, что информация о возможных потерях и высоких рисках отпугнёт новичков.

### 4. Анализ изменений суммы депозитов на платящего пользователя

Новая фича могла повлиять на поведение пользователей.
- Пользователи, которые раньше вносили небольшие суммы, могли стать более осторожными, сократить свои вложения или совсем перестать платить. Это отразится в снижении 25-го перцентиля суммы депозитов в тестовой группе.
- Пользователи, которые склонны к более крупным инвестициям, могли сильнее вовлечься в продукт и начать вносить больше средств. Это отразится в росте 75-го перцентиля.

Используя бутстрап, сравним разницы перцентилей (25, 50, 75) суммы всех депозитов на платящего пользователя в контрольной и тестовой группах:
- Построим доверительные интервалы для разницы между группами для каждого перцентиля.
- Интерпретируем изменения в контексте сформулированной выше гипотезы.
- Визуализируем результат бутстрап-теста. 

Проведем бутстрап-анализ, чтобы оценить 25,50 и 75 перцентили суммы всех депозитов на платящего пользователя в тестовых группах.

```python
np.random.seed(42) # Фиксируем сид для воспроизводимости

# Формируем тестовые группы, считаем сумму всех депозитов на платящего пользователя   
group_control = df_exp.loc[(df_exp['group'] == 'control') & (df_exp['amount'] !=0.0) & \
                           (~df_exp['amount'].isna())].groupby('user_id')['amount'].sum()
group_test = df_exp.loc[(df_exp['group'] == 'test') & (df_exp['amount'] !=0.0) &\
                       (~df_exp['amount'].isna())].groupby('user_id')['amount'].sum()

control = group_control.tolist()
test = group_test.tolist()
```

```python
def boot_strap(group_control, group_teest, n_percentile): 
## Функция получает на вход данные тестовых групп и номер перцентиля для сравнения. 
## Выводит результаты и возвращает массив разниц между перцентилями тестовой и контрольной группы

    n_iterations = 1000 # Задаем количество итераций для бутстрапа
    boot_diffs = []

    for i in range(n_iterations):
        boot_control = np.random.choice(group_control, size=len(group_control), replace=True)
        boot_test = np.random.choice(group_test, size=len(group_test), replace=True)

        # Рассчитываем перцентиль  
        control_p = np.percentile(boot_control, n_percentile)
        test_p = np.percentile(boot_test, n_percentile)
        
        # Находим разницу между значениями тестовой и контрольной групп
        boot_diffs.append(test_p - control_p)

    # Создаём массив разниц между перцентилями тестовой и контрольной групп
    boot_diffs = np.array(boot_diffs)

    # Доверительный интервал разницы между 95-ми перцентилями (перцентильный метод)
    diff_ci = np.percentile(boot_diffs, [2.5, 97.5])

    # Вывод доверительного интервала
    display(f"{n_percentile}%-й перцентиль контрольной группы: {np.percentile(control, n_percentile):.2f}")
    display(f"{n_percentile}%-й перцентиль тестовой группы: {np.percentile(test, n_percentile):.2f}")
    display(f"95%-й доверительный интервал разности {n_percentile}-х перцентилей [{diff_ci[0]:.2f}, {diff_ci[1]:.2f}]")
    return  boot_diffs
```

```python
def boot_hist(boot_diffs, n_percentile):
## Функция получает на вход массив данных из бутстрап-теста и 
## визуализирует гистограмму распределения с 95% доверительным интервалом     
    
    # Доверительный интервал разницы между 95-ми перцентилями (перцентильный метод)
    diff_ci = np.percentile(boot_diffs, [2.5, 97.5])
    
    # Визуализация
    plt.figure(figsize=(9, 5))
    plt.hist(boot_diffs, bins=25, color='#586381', edgecolor='black', alpha=0.99)
  
    plt.axvline(np.mean(boot_diffs), color='#444444', linestyle='solid',
               label=f'Среднее исходной выборки: {np.mean(boot_diffs):.1f}')
    plt.axvline(diff_ci[0], color='#ff7f0e', linestyle='solid',\
               label=f'Нижняя граница ДИ: {diff_ci[0]:.1f}')
    plt.axvline(diff_ci[1], color='#ff7f0e', linestyle='solid',\
               label=f'Верхняя граница ДИ: {diff_ci[1]:.1f}')

    display(f"ДИ: [{diff_ci[0]:.2f}, {diff_ci[1]:.2f}]")
   
    plt.title(f'Бутстрап-распределение {n_percentile} перцентиля')
    plt.ylabel('Сколько раз встречается')
    plt.legend(loc='upper right')
    plt.gca().set_facecolor('white')
    plt.tight_layout()
    plt.show()

    return
```

```python
result = boot_strap(control, test, 25)
```

```python
boot_hist(result,25)
```

```python
result = boot_strap(control, test, 50)
```

```python
boot_hist(result, 50)
```

```python
result = boot_strap(control, test, 75)
```

```python
boot_hist(result, 75)
```

**Вывод:** Анализ бутстрап-распределения 25, 50, 75 процентилей суммы всех депозитов на платящего пользователя показал, что доверительный интервал для разности (test − control) для 25 и 50 протентилей лежит ниже нуля (и не пересекает 0), что указывает на статистически значимое сокращение количества таких пользователей в тестовой группе. 

Доверительный интервал 75 процентиля блольше больше нуля (и не пересекает 0), что говорит об увеличении количества пользователей бОльшей суммой на инвестиции в тестовой группе.

Т.е. в тестовой группе, по сравнению с контрольной, уменьшилось количество пользователей, инвестирующих небольшие суммы, но увеличилось количество пользователей, которые вкладывают суммы, больше средней.
